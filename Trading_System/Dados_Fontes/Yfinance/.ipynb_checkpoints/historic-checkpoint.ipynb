```python
from datetime import date
import pandas as pd
from Alert import Alert 
import yfinance as yf
pd.options.mode.chained_assignment = None  # default='warn'

symbol = "BTC"
period = "1h"

alert = Alert(symbol,period)

if (period == "1h"):
    df = yf.download(tickers=symbol, period='1d', interval=period)
else:
    df = yf.download(tickers=symbol, period='60d', interval=period)

display(df)
print(len(df))
df.index = df.index.tz_localize(None)
#df.to_excel("GGAL_1H_yf.xlsx")
```

Importaçao do historicos diários . O rango maximo depende dos dados disponiveis

```python
BTC = yf.Ticker('BTC-USD')
df = BTC.history(start = '2010-01-01' , end = '2025-10-01')
display(df)
```

```python
df = BTC.history(period = 'max' , interval='5m')
display(df)
```

```python
df_filtrado = df[(df['Dividends'] != 0) | (df['Stock Splits'] != 0)]
display(df_filtrado)
```

```python
import yfinance as yf

ticker = "^BVMBR10"  # ou tente "BR10.BZ", "BR10Y-BR", etc.
ativo = yf.Ticker(ticker)

try:
    info = ativo.info
    print("Nome:", info.get("longName", "Não encontrado"))
    print("Setor:", info.get("sector", "Não disponível"))
except Exception as e:
    print("Ticker não encontrado ou erro:", e)
```
