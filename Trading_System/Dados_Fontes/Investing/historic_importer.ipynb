```python
import pandas as pd

df = pd.read_csv("BR10Y_historico.csv")
display(df.head())
display(len (df))
```

```python
import sqlite3
import pandas as pd

# Converter datas para formato datetime padrão
df["Data"] = pd.to_datetime(df["Data"], format="%d.%m%d").dt.strftime("%Y-%m-%d %H:%M:%S")

# Substituir vírgulas por pontos e converter para float
for col in ["Último", "Abertura", "Máxima", "Mínima"]:
    df[col] = df[col].str.replace(",", ".").astype(float)

# Conectar ao banco de dados
conn = sqlite3.connect("sqtitulosinvesting.db")
cursor = conn.cursor()

# Inserir os dados na tabela
for index, row in df.iterrows():
    cursor.execute("""
        INSERT INTO tbtitulosprezos (
            id_titulosintervalos,
            datetime,
            open,
            high,
            low,
            close,
            volume
        ) VALUES (?, ?, ?, ?, ?, ?, ?)
    """, (
        1,  # id_titulosintervalos fixo
        row["Data"],
        row["Abertura"],
        row["Máxima"],
        row["Mínima"],
        row["Último"],
        None  # volume nulo
    ))

# Salvar e fechar
conn.commit()
conn.close()
```
