rutina que le os dados intradiarios dos titulos mes a mes e insere no BD sqtitulosalpha

```python
from alpha_market import alpha_market_data_intra
from alpha_market import get_or_insert_symbol
from alpha_market import get_or_insert_interval
from alpha_market import get_or_insert_titulos_intervalos
from alpha_market import insert_titulos_prezos
```

```python
#rango de datas a extraer com api 
datainicio = '2002-10'
datafin = '2002-11'
function = 'TIME_SERIES_INTRADAY'
symbol = 'GGAL'
interval = '60min'
apikey = 'BSFBG5DHV3CCXVMV' 
```

```python
from datetime import datetime, timedelta
import time

# Convertendo as strings para objetos datetime
data_inicio = datetime.strptime(datainicio, "%Y-%m")
data_fim = datetime.strptime(datafin, "%Y-%m")

# Inicializando contador de registros
total_registros = 0

# Criando uma lista de meses dentro do intervalo
current_date = data_inicio
while current_date <= data_fim:
    month = current_date.strftime("%Y-%m")
    
    try:       

        # Aguarda 1 segundos antes de cada requisição
        time.sleep(1)
        # Captura via API os dados intradiários correspondentes ao mês
        df = alpha_market_data_intra(function, symbol, interval, month, apikey)
        
        # Insere e obtém os identificadores necessários
        id_titulos = get_or_insert_symbol(symbol)
        id_intervalos = get_or_insert_interval(interval)
        id_titulosintervalos = get_or_insert_titulos_intervalos(id_titulos, id_intervalos)
        
        # Insere os preços e acumula o número de registros inseridos
        inserted = insert_titulos_prezos(id_titulosintervalos, df)
        total_registros += inserted  # Acumulando os registros inseridos

    except Exception as e:
        print(f"Erro ao processar o mês {month}: {str(e)}")

    # Avança para o próximo mês
    current_date += timedelta(days=32)
    current_date = current_date.replace(day=1)

# Exibindo total de registros inseridos ao final
print(f"Total de registros inseridos: {total_registros}")
```
