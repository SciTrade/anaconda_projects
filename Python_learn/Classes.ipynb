```python
import random

class Alert:   
    
    def __init__(self, symbol,intervalo,system):
        self.symbol = symbol
        self.intervalo = intervalo
        self.system = system    

    def mostrar (self):
        display ( f'  Symbol: {self.symbol} ')
        display ( f'  Intervalo: {self.intervalo} ')
        display ( f'  System: {self.system} ')
        display (self.oper())
        
    def oper (self):
        cruce = random.randint(1, 3)
        if cruce == 1:
            orden = 'comprar'
        elif cruce == 2:
            orden = 'mantener'
        else:
            orden = 'vender'
        return orden

Alert_1 = Alert("GGAL","60 min","st3mod")
Alert_2 = Alert('YPF', '5 min', 'stfed')



Alert_1.mostrar()



```
