download um mes completo a partir da data month= (ex: 2009-02)

```python
function = 'TIME_SERIES_INTRADAY'
symbol = 'GGAL'
interval = '60min'
month = '2009-02' 
apikey = 'BSFBG5DHV3CCXVMV' 
```

```python
import requests
import pandas as pd
# Construindo a URL dinamicamente
url = f'https://www.alphavantage.co/query?function={function}&symbol={symbol}&interval={interval}&month={month}&outputsize=full&apikey={apikey}'
r = requests.get(url)
data = r.json()
# Extraindo os dados do JSON
time_series = data.get(f'Time Series ({interval})', {})
# Convertendo para um dataframe do pandas
df = pd.DataFrame.from_dict(time_series, orient='index')

# Renomeando colunas para algo mais intuitivo
df.columns = ['Open', 'High', 'Low', 'Close', 'Volume']

# Adicionando a coluna DateTime
df.insert(0, 'DateTime', df.index)

# Convertendo a coluna DateTime para datetime
df['DateTime'] = pd.to_datetime(df['DateTime'])

# Ajustando o índice para ser a coluna DateTime
df = df.set_index('DateTime')

# Ordenando os dados pelo índice
df = df.sort_index()

# Exibindo o dataframe
display(df)
display(df.dtypes)
```
