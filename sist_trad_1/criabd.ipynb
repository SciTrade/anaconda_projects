```python
import sqlite3

# Conectar ao banco de dados (ou criar se não existir)
conn = sqlite3.connect('sqtitulosalpha.db')
cursor = conn.cursor()

# Criando a tabela tbintervalos
cursor.execute('''
    CREATE TABLE IF NOT EXISTS tbintervalos (
        id_intervalos INTEGER PRIMARY KEY AUTOINCREMENT,
        intervalo TEXT NOT NULL
    )
''')

# Criando a tabela tbtitulos
cursor.execute('''
    CREATE TABLE IF NOT EXISTS tbtitulos (
        id_titulos INTEGER PRIMARY KEY AUTOINCREMENT,
        symbol TEXT NOT NULL,
        descricao TEXT,
        mercado TEXT,
        moeda TEXT
    )
''')

# Criando a tabela tbtitulosintervalos
cursor.execute('''
    CREATE TABLE IF NOT EXISTS tbtitulosintervalos (
        id_titulosintervalos INTEGER PRIMARY KEY AUTOINCREMENT,
        id_titulos INTEGER NOT NULL,
        id_intervalos INTEGER NOT NULL,
        FOREIGN KEY (id_titulos) REFERENCES tbtitulos(id_titulos),
        FOREIGN KEY (id_intervalos) REFERENCES tbintervalos(id_intervalos)
    )
''')

# Criando a tabela tbtitulosprezos
cursor.execute('''
    CREATE TABLE IF NOT EXISTS tbtitulosprezos (
        id_titulosprezos INTEGER PRIMARY KEY AUTOINCREMENT,
        id_titulosintervalos INTEGER NOT NULL,
        datetime TEXT NOT NULL,
        open FLOAT,
        high FLOAT,
        low FLOAT,
        close FLOAT,
        volume FLOAT,
        FOREIGN KEY (id_titulosintervalos) REFERENCES tbtitulosintervalos(id_titulosintervalos)
    )
''')

# Commit e fechar conexão
conn.commit()
conn.close()

print("Banco de dados criado com sucesso!")
```
